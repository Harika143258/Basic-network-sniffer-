```python
from scapy.all import sniff, IP, TCP, UDP, ICMP
import time

def packet_callback(packet):
    # Check if the packet contains an IP Layer
    if packet.haslayer(IP):
        src_ip = packet[IP].src
        dst_ip = packet[IP].dst
        proto = packet[IP].proto

        # Match protocol numbers to human-readable names
        protocol_name = "Unknown"
        if proto == 6:
            protocol_name = "TCP"
        elif proto == 17:
            protocol_name = "UDP"
        elif proto == 1:
            protocol_name = "ICMP"

        print(f"[+] Captured: {src_ip} -> {dst_ip} | Protocol: {protocol_name}")

        # Extract and print raw payload data if it exists
        if packet.haslayer(TCP) and packet[TCP].payload:
            raw_payload = bytes(packet[TCP].payload)[:50]
            print(f"    -> Raw Payload: {raw_payload}")
        elif packet.haslayer(UDP) and packet[UDP].payload:
            raw_payload = bytes(packet[UDP].payload)[:50]
            print(f"    -> Raw Payload: {raw_payload}")

print("="*70)
print("     SNIFFER ACTIVATED: Waiting for 30 network packets...       ")
print("     (Make sure to run Cell 3 right after this to generate traffic!)")
print("="*70)

# Start sniffing
sniff(prn=packet_callback, count=30, store=False)

print("\n[!] Project Complete: Finished capturing 30 packets.")
```

```python
# Background Traffic Generator
import requests
import time

print("[*] Generating background network traffic to feed the sniffer...")

# We make rapid web requests to popular public web services
urls = [
    "https://www.google.com",
    "https://www.github.com",
    "https://www.wikipedia.org"
]

for url in urls:
    try:
        print(f" Sending request to {url}")
        # This triggers TCP handshakes and data transfers that Scapy will catch
        requests.get(url, timeout=2)
        time.sleep(0.5)
    except Exception as e:
        pass

print("[*] Traffic generation sequence completed.")
```
